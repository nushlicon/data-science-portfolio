```python

# Lesson 2: Python Variables and Data Types

product = "Laptop"
price = 1200
quantity = 10
discount = 0.10
in_stock = True

print(product)
print(price)
print(quantity)
print(discount)
print(in_stock)
```

```python

# Step 4: Sales Revenue Calculations

total_revenue = price * quantity

discount_amount = total_revenue * discount

net_revenue = total_revenue - discount_amount

print("Total Revenue:", "$",total_revenue)
print("Discount Amount:", "$",discount_amount)
print("Net Revenue:", "$",net_revenue)
```

```python

# Step 6: Checking Python Data Types

print("Product:", type(product))
print("Price:", type(price))
print("Quantity:", type(quantity))
print("Discount:", type(discount))
print("In Stock:", type(in_stock))
print("Net Revenue:", type(net_revenue))
```

```python

# Step 8: Introduction to Python Lists

products = ["Laptop", "Mouse", "Keyboard"]

prices = [1200, 150, 300]

print("Products:", products)
print("Prices:", prices)
```

```python

# Accessing List Elements

print(products[0])
print(products[1])
print(products[2])

print(f"The {products[0]} Price is ${prices[0]}")
print(f"The {products[1]} Price is ${prices[1]}")
print(f"The {products[2]} Price is ${prices[2]}")
```

```python

# Step 10: Modifying Python Lists

# Add a new product and its price
products.append("Monitor")
prices.append(450)

print("Updated Products:", products)
print("Updated Prices:", prices)
```

```python

# Update the price of Mouse

prices[1] = 175

print("Updated Mouse Price:", prices[1])
```

```python

# Remove Keyboard and its corresponding price

products.pop(2)
prices.pop(2)

print("Remaining Products:", products)
print("Remaining Prices:", prices)
```

```python

# Count the number of products

total_products = len(products)

print("Total Products:", total_products)
```

```python
# Step 11: Python Dictionaries
# A dictionary stores information using keys and values.
# Previously, you used two separate lists:

#products = ["Laptop", "Mouse", "Monitor"]
#prices = [1200, 175, 450]

# Step 11: Introduction to Dictionaries
# 11.1 Create your first dictionary
#In python_fundamentals.ipynb, add a new code cell and enter:

laptop = {
    "product": "Laptop",
    "price": 1200,
    "quantity": 5,
    "in_stock": True
}

print(laptop)
```

```python

# Accessing Dictionary Values

print("Product:", laptop["product"])
print("Price:", laptop["price"])
print("Quantity:", laptop["quantity"])
print("In Stock:", laptop["in_stock"])
```

```python

# Calculate Laptop Sales Revenue

laptop_revenue = laptop["price"] * laptop["quantity"]

print("Product:", laptop["product"])
print("Total Revenue: $", laptop_revenue)
```

```python

# Step 13: Lists of Dictionaries

inventory = [
    {"product": "Laptop", "price": 1200, "quantity": 5},
    {"product": "Mouse", "price": 175, "quantity": 10},
    {"product": "Monitor", "price": 450, "quantity": 3}
]

print(inventory)
```

```python

# Access individual products

print(inventory[0])
print(inventory[1])
print(inventory[2])

# Access a specific value
print("Second Product:", inventory[1]["product"],"$",inventory[1]["price"])
print("Second Product Price:", inventory[1]["price"])
```

```python

# Step 14: Introduction to For Loops

for inventory_item in inventory:
    print(inventory_item["product"])
```

```python

# Display Product Names and Prices

for item in inventory:
    print(item["product"], "- Price: $", item["price"])
```

```python

# Calculate Revenue for Every Product

for item in inventory:
    revenue = item["price"] * item["quantity"]

    print(item["product"], "- Revenue: $", revenue)
```

```python
import matplotlib.pyplot as plt

plt.bar([item["product"] for item in inventory], [item["price"] * item["quantity"] for item in inventory])
plt.title("Revenue breakdown by Product")
plt.xlabel("Product")
plt.ylabel("Revenue ($)")
plt.show()
```

```python

# Step 15: Calculate Total Inventory Revenue

total_revenue = 0

for item in inventory:
    revenue = item["price"] * item["quantity"]
    total_revenue = total_revenue + revenue

print("Total Inventory Revenue: $", total_revenue)
```

```python
# Step 15: Calculate Total Inventory Revenue

total_quantity = 0

for item in inventory:
    total_quantity = total_quantity + item["quantity"]

print("Total Inventory Quantity: ", total_quantity)

```

```python

# Classify Revenue for Every Product

for item in inventory:
    revenue = item["price"] * item["quantity"]

    if revenue >= 5000:
        category = "High Revenue"
    elif revenue >= 2000:
        category = "Medium Revenue"
    else:
        category = "Low Revenue"

    print(f"{item['product']}: ${revenue} - {category}")
```

```python

# Step 17: Introduction to Python Functions

def calculate_revenue(price, quantity):
    revenue = price * quantity
    return revenue

laptop_revenue = calculate_revenue(1200, 5)

print("Laptop Revenue: $", laptop_revenue)
```

```python

# Use the Function for Multiple Products

for item in inventory:
    revenue = calculate_revenue(item["price"], item["quantity"])
    print(f"{item['product']}: ${revenue}")
```
