18.2 Create a dataset with quality problems
Add the following code to the first cell:

```python

# Step 18: Introduction to Data Cleaning

import pandas as pd

sales_data = {
    "Product": [
        "Laptop",
        "Mouse",
        "Keyboard",
        "Laptop",
        "Monitor",
        "Mouse"
    ],
    "Price": [
        1200,
        175,
        300,
        1200,
        None,
        175
    ],
    "Quantity": [
        5,
        10,
        None,
        5,
        3,
        10
    ]
}

df = pd.DataFrame(sales_data)

print(df)
```

What are the problems?
Our dataset contains two types of quality issues.

Missing values: The Keyboard quantity and Monitor price are missing. Pandas represents these missing numeric values as NaN, meaning Not a Number.

Duplicate records: The Laptop and Mouse records each appear twice.

However, in a real sales dataset, repeated products might represent legitimate separate transactions. We should only remove duplicates after confirming that the repeated rows are erroneous.

```python

# Check for Missing Values

print(df.isnull().sum())
```

What does this mean?
- df.isnull() checks each cell for a missing value.
- .sum() counts the missing values in each column.
- The results show that Price and Quantity each contain one missing value.

Step 19: Handling Missing Values with Pandas
Now we'll learn how to clean the dataset without changing the original DataFrame.

There are several ways to handle missing values, including deleting incomplete records, replacing missing values, or investigating the original data source.

For this exercise, we'll use imputation, which means replacing missing values with estimates.
19.1 Create a working copy
In data_cleaning.ipynb, add a new code cell:

```python

# Step 19: Create a Working Copy

df_clean = df.copy()

print(df_clean)
```

Why use .copy()? It lets us clean df_clean while preserving the original df for comparison.
19.2 Replace missing prices using the median
Add another code cell:

```python

# Fill Missing Prices Using the Median

median_price = df_clean["Price"].median()

df_clean["Price"] = df_clean["Price"].fillna(median_price)

print("Median Price:", median_price)
print(df_clean)
```

```python

# Fill Missing Quantities Using the Median

median_quantity = df_clean["Quantity"].median()

df_clean["Quantity"] = df_clean["Quantity"].fillna(median_quantity)

print("Median Quantity:", median_quantity)
print(df_clean)
```

Step 20: Validate your cleaning
Add one final code cell:

```python

# Verify Missing Values Have Been Handled

print(df_clean.isnull().sum())
```

```python

# Compare Missing Values Before and After Cleaning

print("Original Dataset:")
print(df.isnull())

print("\nCleaned Dataset:")
print(df_clean.isnull())
```

```python

# Compare Missing Values Using Numbers

print("Original Dataset Missing Values:")
print(df.isnull().sum())

print("\nCleaned Dataset Missing Values:")
print(df_clean.isnull().sum())
print("Total Missing Values in Cleaned Dataset:")
print(df_clean.isnull().sum().sum())
```

```python

# Calculate Percentage of Missing Price Values

total_rows = len(df)
missing_prices = df["Price"].isnull().sum()

missing_percentage = (missing_prices / total_rows) * 100

print(f"Missing Price Percentage: {missing_percentage:.2f}%")
```

That's because your dataset contains 6 rows and 1 missing price.
What does :.2f mean?
In this statement:
print(f"Missing Price Percentage: {missing_percentage:.2f}%")

- f creates a formatted string.
- {missing_percentage} inserts the calculated value.
- :.2f displays the number with two decimal places.
- % displays the percentage symbol.

```python

# Step 21: Identify Duplicate Records

print(df_clean.duplicated())
print("\nTotal Duplicate Rows:", df_clean.duplicated().sum())
```

Let's examine where they came from and why Pandas identifies them as duplicates.
1. Identify the two duplicate rows
Here is your cleaned dataset:
Index	Product	Price	Quantity	Status
0	Laptop	1200	5	Original
1	Mouse	175	10	Original
2	Keyboard	300	5	Unique
3	Laptop	1200	5	Duplicate of row 0
4	Monitor	300	3	Unique
5	Mouse	175	10	Duplicate of row 1

The duplicate pairs are:
- Rows 0 and 3: Both contain Laptop, price $1,200, and quantity 5.
- Rows 1 and 5: Both contain Mouse, price $175, and quantity 10.
Pandas marks rows 3 and 5 as True because those records have already appeared earlier in the dataset.
2. Why doesn't Pandas mark rows 0 and 1 as duplicates?
Your command:
df_clean.duplicated()

uses the default setting keep="first".

This means Pandas keeps the first occurrence marked as False and marks subsequent identical occurrences as True.

That's why your output is:
0    False
1    False
2    False
3     True
4    False
5     True
dtype: bool


3. Display only the duplicate rows
Let's make Python show the actual records instead of just True and False.

Add a new code cell:

```python

# Display Only Duplicate Rows

duplicate_rows = df_clean[df_clean.duplicated(keep='first')]
# use keep='first' to mark duplicates as True except for the first occurrence
# Use duplicate_rows = df_clean[df_clean.duplicated()]
print(duplicate_rows)
```

```python

# Display All Rows Involved in Duplicate Groups

print(df_clean[df_clean.duplicated(keep=False)])
```

22.1 Create a new cleaned DataFrame
In data_cleaning.ipynb, add a new code cell:

```python

# Step 22: Remove Duplicate Records

df_final = df_clean.drop_duplicates()

print(df_final)
```

```python

# Reset the Index

df_final = df_final.reset_index(drop=True)

print(df_final)
```

Here:
- reset_index() creates a new sequential index.
- drop=True prevents the old index from becoming an additional column.
- df_final = stores the result back in your final DataFrame.

```python

# Validate the Final Dataset

print("Total Records:", len(df_final))
print("Missing Values:", df_final.isnull().sum().sum())
print("Duplicate Rows:", df_final.duplicated().sum())
```
