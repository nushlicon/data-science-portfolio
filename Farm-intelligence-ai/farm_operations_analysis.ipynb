```python

import pandas as pd

# Relative path from the current working directory
file_path = "data/Corn_Grain_Production.csv"

# Read the USDA CSV file
df_corn = pd.read_csv(file_path)

# Display the first five records
df_corn.head()
```

```python
# Display information about the DataFrame

print("Dataset dimensions:", df_corn.shape)

print("\nColumn names:")
print(df_corn.columns.tolist())
```

```python
# Display information about the DataFrame

df_corn.info()
```

```python
# Check for missing values in the DataFrame
df_corn.isnull().sum()
```

```python

print("Original dataset:", df_corn.shape)
```

```python

selected_columns = [
    "Year",
    "Period",
    "State",
    "Data Item",
    "Value"
]

df_selected = df_corn[selected_columns].copy()

print("Selected dataset shape:", df_selected.shape)

df_selected.head()
```

```python
# Filter the DataFrame to include only rows where the "Period" column is equal to "YEAR"
df_selected["Period"] == "YEAR"
```

```python

# Keep only annual production records
df_annual = df_selected[
    df_selected["Period"] == "YEAR"
].copy()

print("Original records:", len(df_selected))
print("Annual records:", len(df_annual))

df_annual.head()

```

```python
# Check the data type of the production values
print("Value column data type:")
print(df_annual["Value"].dtype)

# Display the first five production values
print("\nFirst five values:")
print(df_annual["Value"].head())
```

```python
# Convert strings to numbers
# Step 1: Remove commas
df_annual["Value"] = (
    df_annual["Value"]
    .str.replace(",", "", regex=False)
)

# Step 2: Convert strings into numeric values
df_annual["Value"] = pd.to_numeric(
    df_annual["Value"],
    errors="coerce"
)

# Step 3: Verify the conversion
print("New data type:", df_annual["Value"].dtype)
print("Missing values:", df_annual["Value"].isnull().sum())

df_annual.head()
```

```python

print("Annual records:", len(df_annual))
print("Unique years:", df_annual["Year"].nunique())
print("Value data type:", df_annual["Value"].dtype)
print("Missing production values:", df_annual["Value"].isnull().sum())
```

```python

# Calculate production statistics
total_production = df_annual["Value"].sum()
average_production = df_annual["Value"].mean()
highest_production = df_annual["Value"].max()
lowest_production = df_annual["Value"].min()

print(f"Total across annual observations: {total_production:,.0f} bushels")
print(f"Average annual production: {average_production:,.0f} bushels")
print(f"Highest annual production: {highest_production:,.0f} bushels")
print(f"Lowest annual production: {lowest_production:,.0f} bushels")
```

```python

# Find the index of the highest production value
highest_index = df_annual["Value"].idxmax()

# Retrieve the corresponding record
highest_record = df_annual.loc[highest_index]

print("Highest production year:", highest_record["Year"])
print(f"Production: {highest_record['Value']:,.0f} bushels")
```

```python

# Sort the annual records from oldest to newest
df_annual_sorted = df_annual.sort_values(
    by="Year",
    ascending=True
).reset_index(drop=True)

df_annual_sorted[["Year", "Value"]]
```

```python

import matplotlib.pyplot as plt

# Create the chart
plt.figure(figsize=(10, 5))

plt.plot(
    df_annual_sorted["Year"],
    df_annual_sorted["Value"],
    marker="o",
    linewidth=2
)

# Add labels
plt.title("Washington State Corn Grain Production (2015–2025)")
plt.xlabel("Year")
plt.ylabel("Production (Bushels)")

# Show every year on the horizontal axis
plt.xticks(df_annual_sorted["Year"])

# Add gridlines
plt.grid(True, alpha=0.3)

# Display the chart
plt.show()
```

```python

df_annual_sorted["YoY_Change_Percent"] = (
    df_annual_sorted["Value"].pct_change() * 100
)

df_annual_sorted[
    ["Year", "Value", "YoY_Change_Percent"]
].round(2)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(11, 5))

plt.bar(
    df_annual_sorted["Year"],
    df_annual_sorted["YoY_Change_Percent"]
)

plt.axhline(y=0, color="black", linewidth=1)

plt.title("Washington Corn Production: Year-over-Year Change")
plt.xlabel("Year")
plt.ylabel("Production Change (%)")

plt.xticks(df_annual_sorted["Year"])
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Verify the dataset is ready

print("Dataset shape:", df_annual_sorted.shape)
print("Production data type:", df_annual_sorted["Value"].dtype)
print("Missing values:", df_annual_sorted["Value"].isnull().sum())

df_annual_sorted[["Year", "Value"]]
```

```python

# Calculate the mean and median annual production

mean_production = df_annual_sorted["Value"].mean()
median_production = df_annual_sorted["Value"].median()

print(f"Mean production: {mean_production:,.0f} bushels")
print(f"Median production: {median_production:,.0f} bushels")
```

```python

# Sort production values from smallest to largest

sorted_production = df_annual_sorted["Value"].sort_values()

print(sorted_production.to_string(index=False))
```

```python

# Calculate production variability

variance_production = df_annual_sorted["Value"].var()
std_production = df_annual_sorted["Value"].std()

print(f"Variance: {variance_production:,.0f} bushels squared")
print(f"Standard deviation: {std_production:,.0f} bushels")
```

```python
# Calculate relative production variability

cv_production = (
    std_production / mean_production
) * 100

print(f"Coefficient of variation: {cv_production:.2f}%")
```

```python

# Quartiles and IQR

q1 = df_annual_sorted["Value"].quantile(0.25)
q2 = df_annual_sorted["Value"].quantile(0.50)
q3 = df_annual_sorted["Value"].quantile(0.75)

iqr = q3 - q1

print(f"Q1 (25th percentile): {q1:,.0f} bushels")
print(f"Q2 (Median): {q2:,.0f} bushels")
print(f"Q3 (75th percentile): {q3:,.0f} bushels")
print(f"IQR: {iqr:,.0f} bushels")
```

```python

# Calculate IQR outlier boundaries

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print(f"Lower boundary: {lower_bound:,.0f}")
print(f"Upper boundary: {upper_bound:,.0f}")

print("----------------------------------------")
# Identify potential outliers

outliers = df_annual_sorted[
    (df_annual_sorted["Value"] < lower_bound) |
    (df_annual_sorted["Value"] > upper_bound)
]

print("Number of potential outliers:", len(outliers))

outliers[["Year", "Value"]]

```
